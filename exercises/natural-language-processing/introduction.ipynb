```python
from nltk.corpus import wordnet as wn

synonyms = []
for syn in wn.synsets("car"):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())

print(set(synonyms))
```

```python
text = "this is car"
```

```python
new_sentences = []
for i in synonyms:
    new_sentences.append(text.replace("car", i))

print(new_sentences)
```

```python
from nltk.corpus import wordnet as wn
import random

text = "this is a car and this is a bike"

synonyms = []

words = text.split()
random_word = random.choice(words)
print(f"Random word: {random_word}")
for syn in wn.synsets(random_word):
    for lemma in syn.lemmas():
        synonyms.append(lemma.name())
        new_sentences = []
        for i in synonyms:
            new_sentences.append(text.replace(random_word, i))

print(synonyms)
print(new_sentences)
```

```python
from nltk.corpus import wordnet as wn
import random

text_2 = "Despite multiple complications and her near-death experience"

synonyms_2 = []

words_2 = text_2.split()
random_word_2 = random.choice(words_2)
print(f"Random word: {random_word_2}")
for syn in wn.synsets(random_word_2):
    for lemma in syn.lemmas():
        synonyms_2.append(lemma.name())
        new_sentences_2 = []
        for i in synonyms_2:
            new_sentences_2.append(text_2.replace(random_word_2, i))

print(synonyms_2)
print(new_sentences_2)
```
